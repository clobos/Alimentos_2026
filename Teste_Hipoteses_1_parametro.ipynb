# Teste de Hipóteses para 1 média

```python
import pandas as pd
from scipy import stats

url = ("https://archive.ics.uci.edu/ml/machine-learning-databases"
       "/wine-quality/winequality-red.csv")
df = pd.read_csv(url, sep=";")
ph = df["pH"].dropna()
mu0 = 3.30
# Estatística e valor-p
t_stat, p_value = stats.ttest_1samp(ph, popmean=mu0)
# Graus de liberdade e valor crítico (alfa = 0,05, bilateral)
n  = len(ph)
gl = n - 1
t_crit = stats.t.ppf(0.975, df=gl)
# Valor-p de uma cauda (unilateral direito)
p_one = stats.t.sf(t_stat, df=gl)
print(f"p-valor:{p_one}")
print(f"n={n}, x̄={ph.mean():.4f}, S={ph.std(ddof=1):.4f}")
print(f"t_obs={t_stat:.4f}, t_crit=±{t_crit:.4f}")
print(f"valor-p bilateral={p_value:.4f}")
```

# Teste de Hipóteses para 1 variâncias

```python
import numpy as np
from scipy import stats
ph = df["pH"].dropna()
n = len(ph)
s2 = np.var(ph, ddof=1) # variância amostral
sig0_sq = 0.0225 # referência didática
Q = (n - 1) * s2 / sig0_sq # estatística de teste
gl = n - 1
# Valor-p bilateral (convenção de caudas iguais)
p_lower = stats.chi2.cdf(Q, df=gl)
p_upper = stats.chi2.sf(Q, df=gl)
p_val = min(1.0, 2 * min(p_lower, p_upper))

# Valores críticos
chi2_inf = stats.chi2.ppf(0.025, df=gl)
chi2_sup = stats.chi2.ppf(0.975, df=gl)

print(f"S²={s2:.6f}, Q_obs={Q:.4f}")
print(f"χ²_inf={chi2_inf:.4f}, χ²_sup={chi2_sup:.4f}")
print(f"valor-p bilateral={p_val:.4f}")
```

# Teste de Hipóteses para 1 proporção

```python
from scipy.stats import binomtest, norm
from statsmodels.stats.proportion import (
 proportions_ztest, proportion_confint)

p0 = 0.15
indicador = (df["quality"] >= 7).astype(int)
k = indicador.sum()
n = len(indicador)
p_hat = k / n

# 1. Teste binomial exato
res_exact = binomtest(k, n, p=p0, alternative="two-sided")
print(f"Exato: k={k}, n={n}, p̂={p_hat:.4f}")
print(f"valor-p exato = {res_exact.pvalue:.4f}")

# 2. Verificação das condições para aproximação Z
print(f"n·p0={n*p0:.1f}, n·(1-p0)={n*(1-p0):.1f}")

# 3. Aproximação Z (statsmodels)
z_stat, p_z = proportions_ztest(k, n, value=p0, prop_var=p0)
print(f"Z_obs={z_stat:.4f}, valor-p Z={p_z:.4f}")
```
