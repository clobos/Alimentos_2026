### Teste t para diferença de médias: caso independente

Este teste é utilizado para comparar as médias de duas populações independentes.

* **Hipótese Nula ($H_0$):** As médias das duas populações são iguais.
  $$H_0: \mu_1 = \mu_2$$
* **Hipótese Alternativa ($H_1$):** As médias das duas populações são diferentes.
  $$H_1: \mu_1 \neq \mu_2$$



#### Variâncias iguais (Teste t de Student para duas amostras)

Assume-se que as duas populações possuem a mesma variância original (homocedasticidade).

* **Estatística do Teste ($t$):**
  $$t = \frac{\bar{X}_1 - \bar{X}_2}{s_p \sqrt{\frac{1}{n_1} + \frac{1}{n_2}}}$$

  Onde $s_p$ é o desvio padrão combinado (pooled):
  $$s_p = \sqrt{\frac{(n_1-1)s_1^2 + (n_2-1)s_2^2}{n_1 + n_2 - 2}}$$

* **Graus de Liberdade ($gl$):**
  $$gl = n_1 + n_2 - 2$$

#### Variâncias diferentes (Teste t de Welch)

Não se assume igualdade de variâncias entre os dois grupos (heterocedasticidade).

* **Estatística do Teste ($t$):**
  $$t = \frac{\bar{X}_1 - \bar{X}_2}{\sqrt{\frac{s_1^2}{n_1} + \frac{s_2^2}{n_2}}}$$

* **Graus de Liberdade ($gl$):** Calculados pela aproximação de Welch-Satterthwaite:
  $$gl = \frac{\left(\frac{s_1^2}{n_1} + \frac{s_2^2}{n_2}\right)^2}{\frac{\left(\frac{s_1^2}{n_1}\right)^2}{n_1-1} + \frac{\left(\frac{s_2^2}{n_2}\right)^2}{n_2-1}}$$

```python
df = load_wine(as_frame=True).frame
c0 = df.loc[df["target"] == 0, "alcohol"]
c1 = df.loc[df["target"] == 1, "alcohol"]
aux = stats.ttest_ind(c0, c1, equal_var=False)
print(f"t = {aux.statistic:.4f}, p = {aux.pvalue:.4f}, gl = {aux.df:.4f}")
```

### Teste t para diferença de médias: caso dependente (Pareado)

Este teste é utilizado quando temos observações pareadas, ou seja, as mesmas unidades experimentais são medidas sob duas condições ou representam variáveis relacionadas (como as medidas de fenóis e flavonoides no mesmo vinho).

* **Hipótese Nula ($H_0$):** A média das diferenças entre as observações pareadas é igual a zero.
  $$H_0: \mu_d = 0$$
* **Hipótese Alternativa ($H_1$):** A média das diferenças é diferente de zero.
  $$H_1: \mu_d \neq 0$$

* **Estatística do Teste ($t$):**
  $$t = \frac{\bar{d}}{s_d / \sqrt{n}}$$

  Onde:
  * $\bar{d}$ é a média amostral das diferenças ($d_i = x_i - y_i$).
  * $s_d$ é o desvio padrão amostral das diferenças.
  * $n$ é o número de pares.

* **Graus de Liberdade ($gl$):**
  $$gl = n - 1$$

```python
import pandas as pd
import sklearn
from sklearn.datasets import load_wine
from scipy import stats

df = load_wine(as_frame=True).frame

# Pares completos: fenóis totais e flavonoides medidos no mesmo vinho
x = df["total_phenols"]
y = df["flavanoids"]
n = len(df)

# Teste t pareado
d = x - y
aux = stats.ttest_rel(x, y)
print(f"n pares = {n}")
print(f"t = {aux.statistic:.4f}, p = {aux.pvalue:.4f}")
```

### Passo a Passo: Cálculo do P-Valor para cada caso

Para calcular o p-valor associado a uma estatística $t$ em um teste bicaudal (two-tailed), utilizamos a distribuição t de Student disponível em `scipy.stats.t`. O p-valor é calculado como:
$$p\text{-valor} = 2 \times (1 - F(|t|))$$
Onde $F$ é a Função de Distribuição Acumulada (CDF) da distribuição t com os respectivos graus de liberdade.

#### Caso 1: Teste t independente com Variâncias Iguais

```python
import numpy as np
from scipy import stats

# 1. Obter os dados dos grupos
grupo1 = df.loc[df['target'] == 0, 'alcohol']
grupo2 = df.loc[df['target'] == 1, 'alcohol']

# 2. Calcular tamanhos amostrais (n) e médias
n1, n2 = len(grupo1), len(grupo2)
mean1, mean2 = grupo1.mean(), grupo2.mean()

# 3. Calcular variâncias amostrais
var1, var2 = grupo1.var(ddof=1), grupo2.var(ddof=1)

# 4. Calcular o desvio padrão combinado (pooled)
s_p = np.sqrt(((n1 - 1) * var1 + (n2 - 1) * var2) / (n1 + n2 - 2))

# 5. Calcular a estatística t
t_stat = (mean1 - mean2) / (s_p * np.sqrt(1/n1 + 1/n2))

# 6. Graus de liberdade
gl = n1 + n2 - 2

# 7. Calcular o p-valor (bicaudal)
p_val = 2 * (1 - stats.t.cdf(np.abs(t_stat), df=gl))

print("[Variâncias Iguais]")
print(f"t calculado: {t_stat:.4f}")
print(f"Graus de liberdade: {gl}")
print(f"p-valor calculado: {p_val:.6e}")
```

#### Caso 2: Teste t independente com Variâncias Diferentes (Welch)

```python
# 1. Obter os dados dos grupos
grupo1 = df.loc[df['target'] == 0, 'alcohol']
grupo2 = df.loc[df['target'] == 1, 'alcohol']

# 2. Calcular tamanhos amostrais (n) e médias
n1, n2 = len(grupo1), len(grupo2)
mean1, mean2 = grupo1.mean(), grupo2.mean()

# 3. Calcular variâncias amostrais
var1, var2 = grupo1.var(ddof=1), grupo2.var(ddof=1)

# 4. Calcular a estatística t
t_stat = (mean1 - mean2) / np.sqrt((var1 / n1) + (var2 / n2))

# 5. Calcular os graus de liberdade de Welch-Satterthwaite
num = ((var1 / n1) + (var2 / n2)) ** 2
den = ((var1 / n1) ** 2 / (n1 - 1)) + ((var2 / n2) ** 2 / (n2 - 1))
gl = num / den

# 6. Calcular o p-valor (bicaudal)
p_val = 2 * (1 - stats.t.cdf(np.abs(t_stat), df=gl))

print("[Variâncias Diferentes - Welch]")
print(f"t calculado: {t_stat:.4f}")
print(f"Graus de liberdade: {gl:.4f}")
print(f"p-valor calculado: {p_val:.6e}")
```

#### Caso 3: Teste t pareado (Dependentes)

```python
# 1. Obter as variáveis pareadas
x = df['total_phenols']
y = df['flavanoids']

# 2. Calcular as diferenças individuais
diferencas = x - y

# 3. Obter média, desvio padrão das diferenças e número de pares
n = len(diferencas)
mean_d = diferencas.mean()
s_d = diferencas.std(ddof=1)

# 4. Calcular a estatística t
t_stat = mean_d / (s_d / np.sqrt(n))

# 5. Graus de liberdade
gl = n - 1

# 6. Calcular o p-valor (bicaudal)
p_val = 2 * (1 - stats.t.cdf(np.abs(t_stat), df=gl))

print("[Pareado]")
print(f"t calculado: {t_stat:.4f}")
print(f"Graus de liberdade: {gl}")
print(f"p-valor calculado: {p_val:.6e}")
```
